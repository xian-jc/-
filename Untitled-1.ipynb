根据csv文件绘制相关的图表

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

# 设置中文字体和样式
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
plt.style.use('seaborn-v0_8-darkgrid')

def plot_multi_line_chart_from_csv(csv_file, x_col=None, y_cols=None, title="多行折线图"):
    """
    从CSV文件绘制多行折线图，每行数据用不同颜色
    
    参数:
    csv_file: CSV文件路径
    x_col: X轴数据列名，如果为None则使用索引
    y_cols: Y轴数据列名列表，如果为None则使用所有数值列
    title: 图表标题
    """
    
    # 1. 读取数据
    try:
        df = pd.read_csv(csv_file)
        print(f"成功读取文件: {csv_file}")
        print(f"数据形状: {df.shape}")
        print(f"列名: {list(df.columns)}")
        print(f"\n数据预览:")
        print(df.head())
    except Exception as e:
        print(f"读取文件失败: {e}")
        return
    
    # 2. 自动确定X轴和Y轴列
    if x_col is None:
        # 尝试找到日期/时间列或使用索引
        date_cols = [col for col in df.columns if any(keyword in col.lower() 
                     for keyword in ['date', 'time', 'year', 'month', 'day', '日期', '时间', '年', '月'])]
        if date_cols:
            x_col = date_cols[0]
            print(f"\n自动选择X轴列: {x_col}")
        else:
            # 如果没有日期列，使用索引
            x_col = 'index'
            df[x_col] = range(len(df))
    
    if y_cols is None:
        # 自动选择数值列作为Y轴
        numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
        if x_col in numeric_cols:
            numeric_cols.remove(x_col)
        y_cols = numeric_cols
    
    print(f"\nX轴列: {x_col}")
    print(f"Y轴列: {y_cols}")
    
    # 3. 准备颜色方案
    num_lines = len(y_cols)
    
    # 方案1: 使用tab20色彩映射（最多20种不同颜色）
    if num_lines <= 20:
        colors = plt.cm.tab20(np.linspace(0, 1, num_lines))
    # 方案2: 使用Set3色彩映射
    elif num_lines <= 30:
        colors = plt.cm.Set3(np.linspace(0, 1, num_lines))
    # 方案3: 生成随机但美观的颜色
    else:
        np.random.seed(42)
        colors = plt.cm.rainbow(np.linspace(0, 1, num_lines))
    
    # 4. 创建图形
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), 
                                  gridspec_kw={'height_ratios': [3, 1]})
    
    # 5. 绘制折线图
    lines = []  # 保存线条对象用于图例
    line_names = []  # 保存线条名称
    
    for i, col in enumerate(y_cols):
        # 处理可能存在的缺失值
        y_data = pd.to_numeric(df[col], errors='coerce')
        x_data = df[x_col]
        
        # 绘制线条
        line, = ax1.plot(x_data, y_data, 
                        color=colors[i], 
                        linewidth=2.5,
                        marker='o' if len(df) <= 20 else '',  # 数据点少时加标记点
                        markersize=6,
                        markerfacecolor='white',
                        markeredgewidth=1.5,
                        label=col)
        
        lines.append(line)
        line_names.append(col)
        
        # 添加数据点标签（数据点较少时）
        if len(df) <= 10:
            for x_val, y_val in zip(x_data, y_data):
                if pd.notna(y_val):
                    ax1.annotate(f'{y_val:.2f}', 
                                xy=(x_val, y_val),
                                xytext=(0, 10),
                                textcoords='offset points',
                                ha='center',
                                fontsize=9,
                                color=colors[i])
    
    # 6. 设置主图属性
    ax1.set_title(title, fontsize=16, fontweight='bold', pad=20)
    ax1.set_xlabel(x_col, fontsize=12)
    ax1.set_ylabel('数值', fontsize=12)
    ax1.grid(True, alpha=0.3, linestyle='--')
    
    # 智能设置X轴刻度
    if len(df) > 10:
        # 如果数据点多，旋转标签并减少显示数量
        ax1.set_xticks(ax1.get_xticks()[::max(1, len(df)//10)])
        plt.setp(ax1.get_xticklabels(), rotation=45, ha='right')
    
    # 7. 添加图例（如果线条太多，放在独立图表中）
    if num_lines <= 15:
        ax1.legend(lines, line_names, loc='best', fontsize=10, 
                  framealpha=0.9, shadow=True)
    else:
        # 线条太多时，创建单独的颜色图例
        print(f"\n提示: 有{num_lines}条线，图例将单独显示在下方")
    
    # 8. 创建颜色图例表格
    legend_data = []
    for i, col in enumerate(y_cols[:min(30, len(y_cols))]):  # 最多显示30行
        legend_data.append({
            '指标': col,
            '最新值': f"{df[col].iloc[-1]:.2f}" if len(df) > 0 else 'N/A',
            '颜色': plt.Rectangle((0,0), 1, 1, fc=colors[i])
        })
    
    # 隐藏第二张图的坐标轴
    ax2.axis('tight')
    ax2.axis('off')
    
    # 创建图例表格
    if legend_data:
        table_data = [[d['指标'], d['最新值']] for d in legend_data]
        colors_for_table = [d['颜色'] for d in legend_data]
        
        table = ax2.table(cellText=table_data,
                         colLabels=['指标名称', '最新值'],
                         cellLoc='left',
                         loc='center',
                         colWidths=[0.6, 0.2])
        
        table.auto_set_font_size(False)
        table.set_fontsize(10)
        table.scale(1, 1.5)
        
        # 设置表格样式
        for i in range(len(table_data) + 1):  # +1 包含标题行
            for j in range(2):
                cell = table[i, j]
                if i == 0:  # 标题行
                    cell.set_facecolor('#404040')
                    cell.set_text_props(color='white', weight='bold')
                elif i-1 < len(colors_for_table):
                    cell.set_facecolor(colors_for_table[i-1])
    
    # 9. 调整布局
    plt.suptitle(f'多指标趋势分析\n数据文件: {csv_file}', 
                fontsize=14, fontweight='bold', y=0.95)
    plt.tight_layout()
    
    # 10. 保存图表
    output_file = f"multi_line_chart_{datetime.now().strftime('%Y%m%d_%H%M%S')}.png"
    plt.savefig(output_file, dpi=300, bbox_inches='tight', facecolor='white')
    print(f"\n图表已保存为: {output_file}")
    
    plt.show()
    
    # 11. 返回处理后的数据用于进一步分析
    return df



if __name__ == "__main__":

    your_csv_file = "your_data.csv" 
    df_result = plot_multi_line_chart_from_csv(
        csv_file=your_csv_file,
        x_col=None,  # 自动检测X轴
        y_cols=None,  # 自动选择所有数值列
        title="您的数据趋势分析"
    )
    
    # 额外功能：数据统计摘要
    if df_result is not None:
        print("\n" + "="*50)
        print("数据统计摘要")
        print("="*50)
        
        numeric_cols = df_result.select_dtypes(include=[np.number]).columns
        if len(numeric_cols) > 0:
            stats = df_result[numeric_cols].agg(['min', 'max', 'mean', 'std'])
            print(stats.round(2))
```
